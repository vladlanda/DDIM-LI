```python
import numpy as np
from matplotlib import pyplot as plt
from glob import glob
import os
```

```python
npz_folder = os.path.join('outputs','forecast','h1_dim64')
npz_files = glob(os.path.join(npz_folder,'*.npz'))
file = npz_files[-1]
file
```

```python
data = np.load(file)
print(data)
print(data['channels'])
```

```python
steps,chs,w,h = data['context'].shape
step_index = -1
ir_img = np.zeros((0,w,h))
for i in [0,2,3]:
    ch = data['context'][step_index][i]
    ch = np.expand_dims(ch,axis=0)
    ir_img = np.concatenate((ir_img,ch),axis=0)
li_img = data['context'][step_index][1]
```

```python
%matplotlib qt
plt.subplot(1,2,1)
plt.imshow(ir_img.transpose(1,2,0))
plt.subplot(1,2,2)
plt.imshow(li_img)
plt.show()
```

```python
ens = data['ensemble']
n,p,c,w,h, = ens.shape
ens_index = 0
step_index = 0

res_ir_img = np.zeros((0,w,h))
for i in [0,2,3]:
    ch = ens[ens_index][step_index][i]
    ch = np.expand_dims(ch,axis=0)
    res_ir_img = np.concatenate((res_ir_img,ch),axis=0)
res_li_img = ens[ens_index][step_index][1]
```

```python
%matplotlib qt
plt.subplot(1,2,1)
plt.imshow(res_ir_img.transpose(1,2,0))
plt.subplot(1,2,2)
plt.imshow(res_li_img)
plt.show()
```

```python
%matplotlib qt

pred_ir_img = res_ir_img.transpose(1,2,0) + ir_img.transpose(1,2,0)
pred_li_img = res_li_img + li_img

plt.subplot(1,2,1)
plt.imshow(pred_ir_img)
plt.subplot(1,2,2)
plt.imshow(pred_li_img)
plt.show()
```
